Task 4: Most Preferred Investment Avenue

Objective:
To identify the most preferred investment avenue among participants using data analysis and visualization.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Load dataset
df = pd.read_csv("Data_set 2 - Copy.csv")
```

```python
# Display column names
print(df.columns)
```

```python
# Count investment avenues
investment_counts = df['Investment_Avenues'].value_counts()

# Display counts
print(investment_counts)
```

```python
# Create bar chart
investment_counts.plot(kind='bar')

plt.title("Most Preferred Investment Avenue")
plt.xlabel("Investment Avenue")
plt.ylabel("Count")

plt.show()
```

### Conclusion
The frequency analysis helped identify the most preferred investment avenue among participants. The visualization clearly shows which investment option is chosen the most.
