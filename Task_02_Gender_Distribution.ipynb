Task 2: Gender Distribution

Objective:
To visualize the gender distribution in the dataset using charts.

Importing Required Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
```

Loading the Dataset

```python
df = pd.read_csv('/content/Data_set 2 - Copy.csv')
```

Displaying Column Names

```python
print(df.columns)
```

Counting Gender Values

```python
print(df['gender'].value_counts())
```

Creating Bar Chart for Gender Distribution

```python
df['gender'].value_counts().plot(kind='bar')

plt.title('Gender Distribution')
plt.xlabel('Gender')
plt.ylabel('Count')

plt.show()
```

Creating Pie Chart for Gender Distribution

```python
df['gender'].value_counts().plot(
    kind='pie',
    autopct='%1.1f%%'
)

plt.title('Gender Distribution')
plt.ylabel('')

plt.show()
```

Conclusion:
The gender distribution was successfully visualized using bar and pie charts.
These visualizations helped in understanding the proportion of genders present in the dataset.
