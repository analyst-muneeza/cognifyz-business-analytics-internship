## Task 7: Common Information Sources

### Objective
To analyze the common sources participants rely on for investment information.


```python
# Import library
import pandas as pd
```

```python
# Load dataset
df = pd.read_csv("Data_set 2 - Copy.csv")
```

```python
# Display column names
print(df.columns)
```

```python
# Display information sources
print(df['Source'].value_counts())
```

```python
# Store information sources count
source_counts = df['Source'].value_counts()

# Display results
print(source_counts)
```

```python
# Import visualization library
import matplotlib.pyplot as plt

# Create bar chart
source_counts.plot(kind='bar')

plt.title("Common Information Sources")
plt.xlabel("Sources")
plt.ylabel("Count")

plt.show()
```

### Conclusion
The analysis identified the most common sources participants use for investment-related information. The visualization clearly highlights the sources preferred by the majority of participants.
