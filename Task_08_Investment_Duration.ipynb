## Task 8: Investment Duration

### Objective
To calculate and analyze the average investment duration of participants.

```python
# Import library
import pandas as pd
```

```python
# Load dataset
df = pd.read_csv("Data_set 2 - Copy.csv")
```

```python
# Display column names
print(df.columns)
```

```python
# Display investment duration values
print(df['Duration'].value_counts())
```

```python
# Convert duration values into numerical form
duration_mapping = {
    'Less than 1 year': 0.5,
    '1-3 years': 2,
    '3-5 years': 4,
    'More than 5 years': 6
}

# Create new numerical column
df['Duration_Numeric'] = df['Duration'].map(duration_mapping)

# Display updated data
print(df[['Duration', 'Duration_Numeric']].head())
```

```python
# Calculate average investment duration
average_duration = df['Duration_Numeric'].mean()

# Display result
print("Average Investment Duration:", average_duration)
```

```python
# Import visualization library
import matplotlib.pyplot as plt

# Visualization
df['Duration'].value_counts().plot(kind='bar')

plt.title("Investment Duration")
plt.xlabel("Duration")
plt.ylabel("Count")

plt.show()
```

### Conclusion
The analysis calculated the average investment duration of participants and identified the most common investment duration category. The visualization provides a clear representation of participants’ investment time preferences.
