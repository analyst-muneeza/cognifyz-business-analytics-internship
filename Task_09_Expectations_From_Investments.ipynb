## Task 9: Expectations from Investments

### Objective
To summarize participants' expectations from investments.

```python
# Import library
import pandas as pd
```

```python
# Load dataset
df = pd.read_csv("Data_set 2 - Copy.csv")
```

```python
# Display column names
print(df.columns)
```

```python
# Display expectations data
print(df['Expect'].value_counts())
```

```python
# Store expectations count
expectation_counts = df['Expect'].value_counts()

# Display results
print(expectation_counts)
```

```python
# Import visualization library
import matplotlib.pyplot as plt

# Create bar chart
expectation_counts.plot(kind='bar')

plt.title("Expectations from Investments")
plt.xlabel("Expectations")
plt.ylabel("Count")

plt.show()
```

### Conclusion
The analysis summarized the common expectations participants have from their investments. The results indicate that factors such as returns, capital appreciation, and financial security are key expectations among investors.
