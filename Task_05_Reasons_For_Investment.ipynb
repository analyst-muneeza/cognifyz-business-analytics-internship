## Task 5: Reasons for Investment

### Objective
To analyze and summarize the common reasons behind investment choices made by participants.

```python
# Import library
import pandas as pd
```

```python
# Load dataset
df = pd.read_csv("Data_set 2 - Copy.csv")
```

```python
# Display column names
print(df.columns)
```

```python
# Display reasons for investment
print(df['Reason_Equity'].value_counts())
print(df['Reason_Mutual'].value_counts())
print(df['Reason_Bonds'].value_counts())
print(df['Reason_FD'].value_counts())
```

```python
# Combine all reasons
all_reasons = pd.concat([
    df['Reason_Equity'],
    df['Reason_Mutual'],
    df['Reason_Bonds'],
    df['Reason_FD']
])

# Count frequency of reasons
reason_counts = all_reasons.value_counts()

# Display results
print(reason_counts)
```

```python
# Visualization
reason_counts.plot(kind='bar')

import matplotlib.pyplot as plt

plt.title("Reasons for Investment")
plt.xlabel("Reasons")
plt.ylabel("Frequency")

plt.show()
```

### Conclusion
The analysis highlighted the common reasons behind investment choices. Factors such as returns, safety, and growth were among the major reasons influencing investment decisions.
