## Task 10: Correlation Analysis

### Objective
To explore potential correlations between factors such as age, investment duration, and expected returns.

```python
# Import libraries
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Load dataset
df = pd.read_csv("Data_set 2 - Copy.csv")
```

```python
# Display column names
print(df.columns)
```

```python
# Convert duration values into numerical form
duration_mapping = {
    'Less than 1 year': 0.5,
    '1-3 years': 2,
    '3-5 years': 4,
    'More than 5 years': 6
}

df['Duration_Numeric'] = df['Duration'].map(duration_mapping)

# Display updated data
print(df[['age', 'Duration', 'Duration_Numeric']].head())
```

```python
# Select relevant numerical columns
correlation_data = df[['age', 'Duration_Numeric']]
correlation_data
```

```python
# Calculate correlation
correlation_matrix = correlation_data.corr()

# Display correlation matrix
print(correlation_matrix)
```

```python
# Create scatter plot
plt.scatter(df['age'], df['Duration_Numeric'])

plt.title("Correlation between Age and Investment Duration")
plt.xlabel("Age")
plt.ylabel("Investment Duration")

plt.show()
```

### Conclusion
The correlation analysis helped explore the relationship between age and investment duration. The visualization and correlation matrix provide insights into how investment duration may vary with participants' age.
