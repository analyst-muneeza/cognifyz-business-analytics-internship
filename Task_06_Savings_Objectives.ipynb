## Task 6: Savings Objectives

### Objective
To identify and present the main savings objectives mentioned by participants in the dataset.

```python
# Import library
import pandas as pd
```

```python
# Load dataset
df = pd.read_csv("Data_set 2 - Copy.csv")
```

```python
# Display column names
print(df.columns)
```

```python
# Display savings objectives
print(df['What are your savings objectives?'].value_counts())
```

```python
# Store savings objectives count
savings_objectives = df['What are your savings objectives?'].value_counts()

# Display results
print(savings_objectives)
```

```python
# Visualization
import matplotlib.pyplot as plt

savings_objectives.plot(kind='bar')

plt.title("Savings Objectives")
plt.xlabel("Objectives")
plt.ylabel("Count")

plt.show()
```

### Conclusion
The analysis identified the major savings objectives of participants. Most individuals focused on objectives such as wealth creation, retirement planning, and future financial security.
