Task 3: Descriptive Statistics

Objective:
To calculate and analyze basic statistical measures for numerical columns in the dataset.

```python
import pandas as pd
```

Loading the Dataset

```python
df = pd.read_csv('/content/Data_set 2 - Copy.csv')
```

Displaying Dataset Columns

```python
print(df.columns)
```

Identifying Numerical Columns

```python
numerical_columns = df.select_dtypes(include=['int64', 'float64'])

print(numerical_columns.columns)
```

Calculating Mean Values

```python
print(numerical_columns.mean())
```

Calculating Median Values

```python
print(numerical_columns.median())
```

Calculating Standard Deviation

```python
print(numerical_columns.std())
```

Displaying Complete Descriptive Statistics

```python
print(numerical_columns.describe())
```

Conclusion:
Descriptive statistical analysis was successfully performed on the numerical columns of the dataset.
Measures such as mean, median, and standard deviation helped in understanding the distribution and variation of the data.
